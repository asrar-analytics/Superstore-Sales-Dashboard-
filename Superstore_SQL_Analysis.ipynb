## Project Overview

This project analyzes the Superstore sales dataset using SQL to explore sales performance, profitability, customer behavior, and business trends.

# 1. Setup

```python
# استيراد المكتبات
import pandas as pd
import sqlite3
```

```python
# قراءة ملف البيانات
df = pd.read_csv("/content/Superstore_Comma.csv")
```

```python
# إنشاء قاعدة بيانات SQLite
conn = sqlite3.connect("superstore.db")
```

```python
# حفظ البيانات داخل جدول superstore
df.to_sql(
    "superstore",
    conn,
    if_exists="replace",
    index=False
)
```

# 2. Data Exploration

```python
# عرض أول 10 سجلات
query = """
SELECT *
FROM superstore
LIMIT 10;
"""

pd.read_sql(query, conn)
```

```python
# معرفة عدد الصفوف الموجودة في جدول superstore

query = """
SELECT COUNT(*) AS total_rows
FROM superstore;
"""

pd.read_sql(query, conn)
```

```python
# عرض أسماء الأعمدة وأنواع البيانات في جدول superstore

query = """
PRAGMA table_info(superstore);
"""

pd.read_sql(query, conn)
```

```python
# حساب عدد القيم الفارغة في أهم الأعمدة

query = """
SELECT
COUNT(*) AS total_rows,

COUNT("Order ID") AS order_id,
COUNT("Customer ID") AS customer_id,
COUNT(Sales) AS sales,
COUNT(Quantity) AS quantity,
COUNT(Profit) AS profit

FROM superstore;
"""

pd.read_sql(query, conn)
```

```python
# عرض الصفوف التي تحتوي على قيم مفقودة

query = """
SELECT *
FROM superstore
WHERE Sales IS NULL
   OR Quantity IS NULL
   OR Profit IS NULL;
"""

pd.read_sql(query, conn)
```

```python
# حذف الصفوف التي لا تحتوي على بيانات مالية
#لحفظ التغييرات conn.commit()
query = """
DELETE FROM superstore
WHERE Sales IS NULL
   OR Quantity IS NULL
   OR Profit IS NULL;
"""

conn.execute(query)
conn.commit()

print("تم حذف الصفوف التي تحتوي على قيم مفقودة.")
```

```python
# التحقق من عدد الصفوف بعد الحذف

query = """
SELECT COUNT(*) AS total_rows
FROM superstore;
"""

pd.read_sql(query, conn)
```

3. Sales Analysis

```python
# حساب أهم مؤشرات المبيعات (KPIs)
#سمى الترميز العلمي (Scientific Notation)2.295274e+06 أي حوالي 2.3 مليون

query = """
SELECT

COUNT(DISTINCT "Order ID") AS total_orders,
SUM(Sales) AS total_sales,
SUM(Quantity) AS total_quantity,
SUM(Profit) AS total_profit

FROM superstore;
"""

pd.read_sql(query, conn)
```

```python
#  بدالة ROUND  تقريب النتائج لتكون أوضح في التقرير

query = """
SELECT

COUNT(DISTINCT "Order ID") AS total_orders,

ROUND(SUM(Sales), 2) AS total_sales,

ROUND(SUM(Quantity), 0) AS total_quantity,

ROUND(SUM(Profit), 2) AS total_profit

FROM superstore;
"""

pd.read_sql(query, conn)
```

```python
# حساب إجمالي المبيعات لكل فئة من المنتجات

query = """
SELECT

Category,

ROUND(SUM(Sales), 2) AS total_sales

FROM superstore

GROUP BY Category

ORDER BY total_sales DESC;
"""

pd.read_sql(query, conn)
```

```python
# حساب إجمالي الأرباح لكل فئة من فئات المنتجات

query = """
SELECT
    Category,
    ROUND(SUM(Profit),2) AS total_profit
FROM superstore
GROUP BY Category
ORDER BY total_profit DESC;
"""

pd.read_sql(query, conn)
```

```python
# حساب إجمالي المبيعات لكل منطقة (Region)
# لمعرفة المنطقة التي تحقق أعلى الإيرادات
query = """
SELECT
    Region,
    ROUND(SUM(Sales), 2) AS total_sales
FROM superstore
GROUP BY Region
ORDER BY total_sales DESC;
"""

pd.read_sql(query, conn)
```

```python
# حساب إجمالي الأرباح لكل منطقة
# لمعرفة المنطقة الأكثر تحقيقًا للأرباح.

query = """
SELECT
    Region,
    ROUND(SUM(Profit),2) AS total_profit
FROM superstore
GROUP BY Region
ORDER BY total_profit DESC;
"""

pd.read_sql(query, conn)
```

```python
# حساب إجمالي الأرباح لكل فئة فرعية من المنتجات
# لمعرفة أكثر المنتجات الفرعية تحقيقًا للأرباح.

query = """
SELECT
    "Sub-Category",
    ROUND(SUM(Profit), 2) AS total_profit
FROM superstore
GROUP BY "Sub-Category"
ORDER BY total_profit DESC;
"""

pd.read_sql(query, conn)
```

```python
# حساب متوسط الخصم وإجمالي الأرباح لكل فئة من المنتجات
# لمعرفة هل الفئات ذات الخصومات الأعلى تحقق أرباحًا أقل.

query = """
SELECT

Category,

ROUND(AVG(Discount),2) AS avg_discount,

ROUND(SUM(Profit),2) AS total_profit

FROM superstore

GROUP BY Category

ORDER BY avg_discount DESC;
"""

pd.read_sql(query, conn)
```

```python
# حساب متوسط الخصم وإجمالي الأرباح لكل فئة فرعية
# لمعرفة أي المنتجات الفرعية تحصل على أكبر خصومات
# وهل تحقق هذه المنتجات أرباحًا أم خسائر.
query = """
SELECT
    "Sub-Category",
    ROUND(AVG(Discount),2) AS avg_discount,
    ROUND(SUM(Profit),2) AS total_profit
FROM superstore
GROUP BY "Sub-Category"
ORDER BY avg_discount DESC;
"""

pd.read_sql(query, conn)
```

```python
# إيجاد أعلى قيمة مبيعات في قاعدة البيانات.
query = """
SELECT
    MAX(Sales) AS highest_sale
FROM superstore;
"""

pd.read_sql(query, conn)
```

# 6.  Time Analysis

```python
# حساب إجمالي المبيعات لكل سنة.

query = """
SELECT
    SUBSTR("Order Date", -4) AS order_year,
    ROUND(SUM(Sales),2) AS total_sales
FROM superstore
GROUP BY order_year
ORDER BY order_year;
"""

pd.read_sql(query, conn)
```

7.  Key Business Insights

```python
# عرض الفئات التي تجاوزت أرباحها 100000
query = """

SELECT
    Category,
    ROUND(SUM(Profit),2) AS total_profit
FROM superstore
GROUP BY Category
HAVING SUM(Profit) > 100000
ORDER BY total_profit DESC;
 """
pd.read_sql(query, conn)
```

```python
# عرض المناطق التي تجاوزت مبيعاتها 600000.

query = """
SELECT
    Region,
    ROUND(SUM(Sales),2) AS total_Sales
FROM superstore
GROUP BY Region
HAVING SUM(Sales) > 600000
ORDER BY total_Sales DESC;
"""

pd.read_sql(query, conn)
```

```python
# عرض المناطق التي تجاوزت مبيعاتها 300000
# داخل فئة Technology فقط.

query = """
SELECT
    Region,
    ROUND(SUM(Sales),2) AS total_sales
FROM superstore

WHERE Category = 'Technology'

GROUP BY Region

HAVING SUM(Sales) > 200000
ORDER BY total_sales DESC;
"""

pd.read_sql(query, conn)
```

تم عرض أول 10 طلبات من فئة Technology في منطقة West فقط. يوضح هذا الاستعلام كيفية استخدام AND لتطبيق أكثر من شرط في نفس الوقت، مما يساعد على تحليل بيانات جزء محدد من الأعمال.

```python
# عرض طلبات فئة Technology الموجودة في منطقة West فقط
query = """
SELECT
    "Order ID",
    "Customer Name",
    Region,
    Category,
    Sales
FROM superstore

WHERE Category = 'Technology'
AND Region = 'West'

LIMIT 10;
"""

pd.read_sql(query, conn)
```

تم عرض الطلبات الموجودة في منطقتي West أو East باستخدام OR. يوضح هذا الاستعلام كيفية دمج أكثر من شرط بحيث يكفي تحقق أحدهما لإظهار السجل.

```python
# عرض الطلبات الموجودة في منطقة West أو منطقة East
query = """

SELECT
    "Order ID","Customer Name",Region,   Sales
FROM superstore
WHERE Region = 'West'
or Region = 'East'
ORDER BY Region DESC;
"""
pd.read_sql(query, conn)
```
